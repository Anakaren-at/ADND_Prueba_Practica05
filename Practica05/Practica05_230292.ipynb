```python
def bar_hor(df, col, title, color, w=None, h=None, lm=0, limit=100, return_trace=False, rev=False):
    cnt_srs = df[col].value_counts()
    yy = cnt_srs.head(limit).index[::-1]
    xx = cnt_srs.head(limit).values[::-1]

    if rev:
        yy = cnt_srs.tail(limit).index[::-1]
        xx = cnt_srs.tail(limit).values[::-1]

    trace = go.Bar(y=yy, x=xx, orientation='h',
                   marker=dict(color=color))

    if return_trace:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_hor_noagg(x, y, title, color, w=None, h=None, lm=0, limit=100, rt=False):
    trace = go.Bar(y=x, x=y, orientation='h',
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=0, rt=False):
    trace = go.Bar(y=y, x=x,
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


trace1 = bar_hor(
    donors_df,
    'Donor City',
    "Ciudades con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace2 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace4 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el menor número de donantes",
    '#c2d2ed',
    600, 400, 200,
    limit=10,
    return_trace=True,
    rev=True
)

fig = tools.make_subplots(
    rows=1,
    cols=2,
    print_grid=False,
    subplot_titles=[
        'Estados con el mayor número de donantes',
        'Estados con el menor número de donantes'
    ]
)

fig.append_trace(trace2, 1, 1)
fig.append_trace(trace4, 1, 2)

fig['layout'].update(height=400, showlegend=False)
iplot(fig)
```

```python
# Crear tabla de contingencia entre estado del proyecto y año de publicación
year_tab = pd.crosstab(projects_df["Estado Actual del Proyecto"], projects_df["Año de Publicación"]).reset_index()

from itertools import zip_longest

# Limpiar la columna de costo (eliminar comas y signo de dólar)
projects_df['costo'] = projects_df['Costo del Proyecto'].apply(lambda x: float(str(x).replace(",", "").replace("$", "")))

# Función para crear gráfica de burbujas (bubble chart)
def cbp(df, col1, col2, aggcol, func, title, cs, bottom_margin=None):
    # Agrupar por las dos columnas y calcular la agregación
    tempdf = df.groupby([col1, col2]).agg({aggcol: func}).reset_index()
    tempdf[aggcol] = tempdf[aggcol].apply(lambda x: int(x))
    tempdf = tempdf.sort_values(aggcol, ascending=False)

    # Crear tamaños de marcadores escalonados
    sizes = list(reversed([i for i in range(10, 31)]))
    intervals = int(len(tempdf) / len(sizes))
    size_array = [9] * len(tempdf)
    
    st = 0
    for i, size in enumerate(sizes):
        for j in range(st, st + intervals):
            size_array[j] = size 
        st = st + intervals
    
    tempdf['size_n'] = size_array
    cols = list(tempdf['size_n'])

    # Crear gráfica de dispersión con marcadores de diferentes tamaños
    trace1 = go.Scatter(
        x=tempdf[col1], 
        y=tempdf[col2], 
        mode='markers', 
        text=tempdf[aggcol],
        marker=dict(size=tempdf.size_n, color=cols, colorscale=cs)
    )
    
    data = [trace1]
    
    if bottom_margin:
        layout = go.Layout(title=title, margin=dict(b=150))
    else:
        layout = go.Layout(title=title)
    
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)

# Llamar la función para graficar costo promedio por mes y año
cbp(projects_df, 'Mes de Publicación', 'Año de Publicación', 'costo', 'mean', 'Costo Promedio de Proyectos por Mes y Año', 'Picnic')
```

## <a id="8.1">8.1 Limpieza de Texto</a>

Escribamos una función utilitaria que normalice el texto y elimine el ruido. Esta función de limpieza de texto realiza la eliminación de palabras vacías (*stopwords*) del texto y la eliminación de signos de puntuación.

```python
# ==========================================
# SECCIÓN 8.1: Limpieza de Texto (Todo en uno)
# ==========================================

# PASO 1: Instalar librerías necesarias (si no están instaladas)
!pip install textblob nltk --quiet

# PASO 2: Importar librerías
from nltk.corpus import stopwords 
import nltk
import string 

# PASO 3: Descargar recursos de NLTK (stopwords)
nltk.download('stopwords', quiet=True)

# Cargar stopwords en inglés (los textos de los proyectos están en inglés)
stop_words = stopwords.words("english")

# PASO 4: Definir la función de limpieza de texto
def _clean(txt):
    """
    Función de limpieza de texto que normaliza y elimina ruido.
    Realiza: conversión a minúsculas, eliminación de puntuación, 
    eliminación de stopwords y palabras específicas de ruido.
    """
    if not isinstance(txt, str):
        return ""
        
    # Convertir a minúsculas
    txt = txt.lower()

    # Eliminar puntuación 
    txt = ''.join(x for x in txt if x not in string.punctuation)

    # Eliminar stopwords
    clean_txt = ""
    for word in txt.split():
        if word in stop_words:
            continue 
        clean_txt += " " + word 
    
    # Eliminar palabras específicas de ruido
    noise = ['students', 'my', 'title', 'would', 'every', 'will', 'many', 'also', 'donotremoveessaydivider']
    for ns in noise:
        clean_txt = clean_txt.replace(ns, "")

    return clean_txt

# PASO 5: Renombrar columnas necesarias (si no están ya en español)
projects_df = projects_df.rename(columns={
    'School ID': 'ID de Escuela',
    'Posted Year': 'Año de Publicación',
    'Posted Month': 'Mes de Publicación',
    'Project Essay': 'Ensayo del Proyecto',
    'Project Need Statement': 'Declaración de Necesidad'
})

schools_df = schools_df.rename(columns={
    'School ID': 'ID de Escuela'
})

# PASO 6: Unir proyectos con escuelas
projects_df = projects_df.merge(schools_df, on='ID de Escuela', how='inner')

# PASO 7: Filtrar proyectos de abril 2018
april18_df = projects_df[(projects_df['Año de Publicación'] == 2018) & (projects_df['Mes de Publicación'] == 4)].copy()

# PASO 8: Aplicar limpieza de texto a las columnas de ensayo y necesidad
print("🧹 Limpiando textos...")
april18_df['clean_essay'] = april18_df['Ensayo del Proyecto'].apply(_clean)
april18_df['clean_need'] = april18_df['Declaración de Necesidad'].apply(_clean)

print("✅ ¡Limpieza completada!")
print(f"Total de proyectos de abril 2018: {len(april18_df)}")

# Mostrar ejemplo de texto original vs limpio
print("\n--- Ejemplo de texto original vs limpio ---")
print("ORIGINAL:", april18_df['Ensayo del Proyecto'].iloc[0][:100], "...")
print("LIMPIO:  ", april18_df['clean_essay'].iloc[0][:100], "...")
```

## <a id="8.2"> 8.2 Análisis de Bolsa de Palabras (Bag of Words) </a>

```python
# ==========================================
# SECCIÓN 8.2.1: Análisis de Bolsa de Palabras (Bag of Words)
# ==========================================

from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns

# Función para generar n-gramas a partir de una lista de palabras
def generate_ngrams(txt, N):
    grams = [txt[i:i+N] for i in range(len(txt)-N+1)]
    grams = [" ".join(b) for b in grams]
    return grams 

# Generar unigramas, bigramas y trigramas a partir del ensayo limpio
april18_df['unigrams'] = april18_df['clean_essay'].apply(lambda x: generate_ngrams(x.split(), 1))
april18_df['bigrams'] = april18_df['clean_essay'].apply(lambda x: generate_ngrams(x.split(), 2))
april18_df['trigrams'] = april18_df['clean_essay'].apply(lambda x: generate_ngrams(x.split(), 3))

# --- Procesamiento de Unigramas (Palabras individuales) ---
all_unigrams = []
for each in april18_df['unigrams']:
    all_unigrams.extend(each)
    
t = Counter(all_unigrams).most_common(25)
x = [a[0] for a in t]  # Palabras
y = [a[1] for a in t]  # Frecuencias

# --- Procesamiento de Bigramas (Pares de palabras) ---
all_bigrams = []
for each in april18_df['bigrams']:
    all_bigrams.extend(each)
    
t1 = Counter(all_bigrams).most_common(25)
x1 = [a[0] for a in t1]  # Bigramas
y1 = [a[1] for a in t1]  # Frecuencias

# --- Procesamiento de Trigramas (Tres palabras consecutivas) ---
all_trigrams = []
for each in april18_df['trigrams']:
    all_trigrams.extend(each)
    
t2 = Counter(all_trigrams).most_common(25)
x2 = [a[0] for a in t2]  # Trigramas
y2 = [a[1] for a in t2]  # Frecuencias

# ==========================================
# Visualización de los resultados
# ==========================================
sns.set_style("dark")
fig, axes = plt.subplots(nrows=1, ncols=5, squeeze=True, figsize=(15, 10))

# Gráfica de Unigramas
bar = sns.barplot(y=x, x=y, ax=axes[0], palette='GnBu_d', edgecolor="white")
bar.set(xlabel='', xticks=[])
axes[0].set_title("Principales palabras clave utilizadas por los maestros", fontsize=12)

# Eliminar ejes vacíos para mejor espaciado
fig.delaxes(axes[1])

# Gráfica de Bigramas
bar1 = sns.barplot(y=x1, x=y1, ax=axes[2], palette='GnBu_d', edgecolor="white")
bar1.set(xlabel='', xticks=[])
axes[2].set_title("Principales bigramas utilizados por los maestros", fontsize=12)

# Eliminar ejes vacíos para mejor espaciado
fig.delaxes(axes[3])

# Gráfica de Trigramas
bar2 = sns.barplot(y=x2, x=y2, ax=axes[4], palette='GnBu_d', edgecolor="white")
bar2.set(xlabel='', xticks=[])
axes[4].set_title("Principales trigramas utilizados por los maestros", fontsize=12)

plt.tight_layout()
plt.show()
```

> Las palabras clave más utilizadas en todos los ensayos, en todos los estados y en todos los tipos de escuelas son <b>"escuela", "aprendizaje", "aula", "lectura"</b>. Mientras que los bigramas se centran principalmente en <b>"asientos flexibles"</b> y <b>"idioma inglés"</b>. ## <a href="9.1"> 9.1 Proyectos Expirados vs Proyectos Financiados </a> 

En esta sección, exploremos y comprendamos las características clave de los proyectos que fueron financiados con éxito y comparémoslos con los proyectos que expiraron. Me centraré en los proyectos del año pasado, es decir, 2017.

```python
# ==========================================
# SECCIÓN 9.1: Proyectos Expirados vs Proyectos Financiados (2017)
# ==========================================

import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==========================================
# PASO 1: Filtrar proyectos de 2017
# IMPORTANTE: Los valores de estado están en INGLÉS
# ==========================================
expirados = projects_df[(projects_df['Año de Publicación'] == 2017) & 
                        (projects_df['Estado Actual del Proyecto'] == 'Expired')].copy()

financiados = projects_df[(projects_df['Año de Publicación'] == 2017) & 
                          (projects_df['Estado Actual del Proyecto'] == 'Fully Funded')].copy()

print(f"✅ Proyectos Expirados (2017): {len(expirados)}")
print(f"✅ Proyectos Financiados (2017): {len(financiados)}")

# ==========================================
# PASO 2: Unir con la tabla de maestros
# ==========================================
expirados = expirados.merge(teachers_df, on="ID de Maestro", how='left')
financiados = financiados.merge(teachers_df, on="ID de Maestro", how='left')


# ==========================================
# PASO 3: Función auxiliar para generar trazas
# ==========================================
def obtener_traza(col, func, flag=None, leg=False):
    t1 = expirados[col].value_counts()
    t2 = financiados[col].value_counts()

    if flag == 'num':
        t1 = expirados[expirados['NumProjects'] <= 20][col].value_counts()    
        t2 = financiados[financiados['NumProjects'] <= 20][col].value_counts() 
    
    trace1 = func(
        x=t1.index,
        y=[float(a)*100/sum(t1.values) for a in t1.values],
        name='Expirado',
        marker=dict(color='#e82041'),
        showlegend=leg,
        opacity=0.9
    )
    trace2 = func(
        x=t2.index,
        y=[float(a)*100/sum(t2.values) for a in t2.values],
        name='Financiado',
        marker=dict(color='#0de286'),
        opacity=0.9, 
        showlegend=leg
    )
    return trace1, trace2


# ==========================================
# PASO 4: Generar trazas
# ==========================================
tr1, tr2 = obtener_traza('Categoría de Nivel de Grado del Proyecto', go.Scatter)
tr3, tr4 = obtener_traza('Tipo de Zona Metropolitana', go.Scatter)
tr7, tr8 = obtener_traza('Categoría de Recursos del Proyecto', go.Bar)
tr9, tr10 = obtener_traza('Mes de Publicación', go.Bar, leg=True)


# ==========================================
# PASO 5: Análisis de Cantidad de Palabras en el Ensayo
# ==========================================
expirados['conteo_palabras'] = expirados['Ensayo del Proyecto'].apply(lambda x: len(str(x).split()))
financiados['conteo_palabras'] = financiados['Ensayo del Proyecto'].apply(lambda x: len(str(x).split()))

bins_wc = [0, 50, 100, 150, 200, 250, 300, 350, 400, 450, 500, 10000]

# Expirados
expirados['wc_bin'] = pd.cut(expirados['conteo_palabras'], bins_wc)
t1_wc = dict(expirados['wc_bin'].value_counts())
labels1 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1_wc.keys()]
vals1 = [float(a)*100/sum(t1_wc.values()) for a in list(t1_wc.values())]
d1 = {labels1[i]: vals1[i] for i in range(len(labels1))}
order_wc = ['0-50', '50-100', '100-150', '150-200', '200-250', '250-300', '300-350', '350-400', '400-450', '450-500', '500-10000']
values1 = [d1.get(v, 0) for v in order_wc]

tr13 = go.Bar(x=order_wc, y=values1, name='Expirado', marker=dict(color='#e82041'), opacity=0.9)

# Financiados
financiados['wc_bin'] = pd.cut(financiados['conteo_palabras'], bins_wc)
t2_wc = dict(financiados['wc_bin'].value_counts())
labels2 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t2_wc.keys()]
vals2 = [float(a)*100/sum(t2_wc.values()) for a in list(t2_wc.values())]
d2 = {labels2[i]: vals2[i] for i in range(len(labels2))}
values2 = [d2.get(v, 0) for v in order_wc]

tr14 = go.Bar(x=order_wc, y=values2, name='Financiado', marker=dict(color='#0de286'), opacity=0.9)


# ==========================================
# PASO 6: Análisis de Costo del Proyecto
# Usamos la columna 'costo' que ya existe y está limpia
# ==========================================
bins_cost = [0, 100, 500, 1000, 1500, 2000, 2500, 3000, 70000]
order_cost = ['100-500', '500-1000', '1000-1500', '1500-2000', '2000-2500', '2500-3000', '3000-70000']

# Expirados
expirados['costo_bin'] = pd.cut(expirados['costo'], bins_cost)
t1_cost = dict(expirados['costo_bin'].value_counts())
labels1_c = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1_cost.keys()]
vals1_c = [float(a)*100/sum(t1_cost.values()) for a in list(t1_cost.values())]
d1_c = {labels1_c[i]: vals1_c[i] for i in range(len(labels1_c))}
values1_c = [d1_c.get(v, 0) for v in order_cost]

tr15 = go.Scatter(x=order_cost, y=values1_c, name='Expirado', fill='tozeroy', marker=dict(color='#e82041'), opacity=0.9)

# Financiados
financiados['costo_bin'] = pd.cut(financiados['costo'], bins_cost)
t2_cost = dict(financiados['costo_bin'].value_counts())
labels2_c = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t2_cost.keys()]
vals2_c = [float(a)*100/sum(t2_cost.values()) for a in list(t2_cost.values())]
d2_c = {labels2_c[i]: vals2_c[i] for i in range(len(labels2_c))}
values2_c = [d2_c.get(v, 0) for v in order_cost]

tr16 = go.Scatter(x=order_cost, y=values2_c, name='Financiado', fill='tozeroy', marker=dict(color='#0de286'), opacity=0.9)


# ==========================================
# PASO 7: Crear el Dashboard con Subplots
# ==========================================
titulos_graficas = [
    'Categoría de Nivel de Grado', 
    "Tipo de Zona Metropolitana", 
    "Costo del Proyecto (en $)", 
    "Cantidad de Palabras en el Ensayo", 
    "Mes de Publicación", 
    "Categoría de Recursos"
]

fig = make_subplots(
    rows=3, cols=2, 
    subplot_titles=titulos_graficas, 
    specs=[[{}, {}], [{},{}], [{}, {}]]
) 

fig.append_trace(tr1, 1, 1)
fig.append_trace(tr2, 1, 1)
fig.append_trace(tr3, 1, 2)
fig.append_trace(tr4, 1, 2)

fig.append_trace(tr15, 2, 1) # Costo
fig.append_trace(tr16, 2, 1)

fig.append_trace(tr13, 2, 2) # Palabras
fig.append_trace(tr14, 2, 2)

fig.append_trace(tr9, 3, 1)  # Mes
fig.append_trace(tr10, 3, 1)

fig.append_trace(tr7, 3, 2)  # Recursos
fig.append_trace(tr8, 3, 2)

# Actualizar el diseño general
fig.update_layout(
    barmode='group',
    title={
        'text': 'Comparación de Características Clave: <b>Proyectos Expirados vs Financiados</b>',
        'font': {'size': 22, 'color': '#000'},
        'y': 0.98,
        'x': 0.5,
        'xanchor': 'center'
    },
    margin=dict(t=100, b=100),
    paper_bgcolor='rgb(254, 247, 234)',
    plot_bgcolor='rgb(254, 247, 234)',
    height=1000
)

# Mostrar la figura
fig.show()
```

```python
# ==========================================
# SECCIÓN 9.2: Proyectos con Duración de Financiación Corta vs Larga
# ==========================================

import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==========================================
# PASO 1: Preparar columna de tiempo de financiación
# ==========================================
# La columna 'Tiempo de Financiamiento' contiene valores como "X days" o "NaT"
# Extraemos solo el número de días
projects_df['tiempo_dias'] = projects_df['Tiempo de Financiamiento'].apply(
    lambda x: str(x).split()[0] if pd.notna(x) else 'NaT'
)

# Filtrar proyectos con tiempo válido y convertir a entero
tempdf = projects_df[projects_df['tiempo_dias'] != "NaT"].copy()
tempdf['tiempo_dias'] = tempdf['tiempo_dias'].apply(int)

# Filtrar solo proyectos de 2017
tempdf = tempdf[tempdf['Año de Publicación'] == 2017]

print(f"Total proyectos de 2017 con tiempo de financiación válido: {len(tempdf)}")

# ==========================================
# PASO 2: Crear dos grupos de comparación
# ==========================================
# df1: Financiación rápida (≤ 1 día)
df1 = tempdf[tempdf['tiempo_dias'] <= 1].copy()

# df2: Financiación lenta (≥ 60 días)
df2 = tempdf[tempdf['tiempo_dias'] >= 60].copy()

print(f"Proyectos con financiación ≤ 1 día: {len(df1)}")
print(f"Proyectos con financiación ≥ 60 días: {len(df2)}")

# ==========================================
# PASO 3: Unir con tabla de maestros
# ==========================================
df1 = df1.merge(teachers_df, on="ID de Maestro", how='left')
df2 = df2.merge(teachers_df, on="ID de Maestro", how='left')


# ==========================================
# PASO 4: Función auxiliar para generar trazas
# ==========================================
def obtener_traza(col, func, flag=None, leg=False):
    t1 = df1[col].value_counts()
    t2 = df2[col].value_counts()

    if flag == 'num':
        t1 = df1[df1['NumProjects'] <= 20][col].value_counts()    
        t2 = df2[df2['NumProjects'] <= 20][col].value_counts() 
    
    trace1 = func(
        x=t1.index,
        y=[float(a)*100/sum(t1.values) for a in t1.values],
        name='Financiación Rápida (≤1 día)',
        marker=dict(color='#0de286'),
        showlegend=leg,
        opacity=0.9
    )
    trace2 = func(
        x=t2.index,
        y=[float(a)*100/sum(t2.values) for a in t2.values],
        name='Financiación Lenta (≥60 días)',
        marker=dict(color='#e82041'),
        opacity=0.9, 
        showlegend=leg
    )
    return trace1, trace2


# ==========================================
# PASO 5: Generar trazas para las diferentes categorías
# ==========================================
tr1, tr2 = obtener_traza('Categoría de Nivel de Grado del Proyecto', go.Scatter)
tr3, tr4 = obtener_traza('Tipo de Zona Metropolitana', go.Scatter)
tr7, tr8 = obtener_traza('Categoría de Recursos del Proyecto', go.Bar)
tr9, tr10 = obtener_traza('Mes de Publicación', go.Bar, leg=True)


# ==========================================
# PASO 6: Análisis de Cantidad de Palabras en el Ensayo
# ==========================================
df1['conteo_palabras'] = df1['Ensayo del Proyecto'].apply(lambda x: len(str(x).split()))
df2['conteo_palabras'] = df2['Ensayo del Proyecto'].apply(lambda x: len(str(x).split()))

bins_wc = [0, 50, 100, 150, 200, 250, 300, 350, 400, 450, 500, 10000]

# Financiación rápida
df1['wc_bin'] = pd.cut(df1['conteo_palabras'], bins_wc)
t1_wc = dict(df1['wc_bin'].value_counts())
labels1 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1_wc.keys()]
vals1 = [float(a)*100/sum(t1_wc.values()) for a in list(t1_wc.values())]
d1 = {labels1[i]: vals1[i] for i in range(len(labels1))}
order_wc = ['0-50', '50-100', '100-150', '150-200', '200-250', '250-300', '300-350', '350-400', '400-450', '450-500', '500-10000']
values1 = [d1.get(v, 0) for v in order_wc]

tr13 = go.Bar(x=order_wc, y=values1, name='Financiación Rápida (≤1 día)', 
              marker=dict(color='#0de286'), showlegend=False, opacity=0.9)

# Financiación lenta (CORREGIDO: usa t2_wc en lugar de t1_wc)
df2['wc_bin'] = pd.cut(df2['conteo_palabras'], bins_wc)
t2_wc = dict(df2['wc_bin'].value_counts())
labels2 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t2_wc.keys()]
vals2 = [float(a)*100/sum(t2_wc.values()) for a in list(t2_wc.values())]
d2 = {labels2[i]: vals2[i] for i in range(len(labels2))}
values2 = [d2.get(v, 0) for v in order_wc]

tr14 = go.Bar(x=order_wc, y=values2, name='Financiación Lenta (≥60 días)', 
              marker=dict(color='#e82041'), showlegend=False, opacity=0.9)


# ==========================================
# PASO 7: Análisis de Costo del Proyecto
# ==========================================
bins_cost = [0, 100, 500, 1000, 1500, 2000, 2500, 3000, 70000]
order_cost = ['100-500', '500-1000', '1000-1500', '1500-2000', '2000-2500', '2500-3000', '3000-70000']

# Financiación rápida
df1['costo_bin'] = pd.cut(df1['costo'], bins_cost)
t1_cost = dict(df1['costo_bin'].value_counts())
labels1_c = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1_cost.keys()]
vals1_c = [float(a)*100/sum(t1_cost.values()) for a in list(t1_cost.values())]
d1_c = {labels1_c[i]: vals1_c[i] for i in range(len(labels1_c))}
values1_c = [d1_c.get(v, 0) for v in order_cost]

tr15 = go.Scatter(x=order_cost, y=values1_c, name='Financiación Rápida (≤1 día)', 
                  fill='tozeroy', marker=dict(color='#0de286'), showlegend=False, opacity=0.9)

# Financiación lenta (CORREGIDO: usa t2_cost en lugar de t1_cost)
df2['costo_bin'] = pd.cut(df2['costo'], bins_cost)
t2_cost = dict(df2['costo_bin'].value_counts())
labels2_c = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t2_cost.keys()]
vals2_c = [float(a)*100/sum(t2_cost.values()) for a in list(t2_cost.values())]
d2_c = {labels2_c[i]: vals2_c[i] for i in range(len(labels2_c))}
values2_c = [d2_c.get(v, 0) for v in order_cost]

tr16 = go.Scatter(x=order_cost, y=values2_c, name='Financiación Lenta (≥60 días)', 
                  fill='tozeroy', marker=dict(color='#e82041'), showlegend=False, opacity=0.9)


# ==========================================
# PASO 8: Crear el Dashboard con Subplots
# ==========================================
titulos_graficas = [
    'Categoría de Nivel de Grado', 
    "Tipo de Zona Metropolitana", 
    "Costo del Proyecto (en $)", 
    "Cantidad de Palabras en el Ensayo", 
    "Mes de Publicación", 
    "Categoría de Recursos"
]

fig = make_subplots(
    rows=3, cols=2, 
    subplot_titles=titulos_graficas, 
    specs=[[{}, {}], [{},{}], [{}, {}]]
) 

# Agregar trazas a las subgráficas (Fila, Columna)
fig.append_trace(tr1, 1, 1)
fig.append_trace(tr2, 1, 1)
fig.append_trace(tr3, 1, 2)
fig.append_trace(tr4, 1, 2)

fig.append_trace(tr15, 2, 1) # Costo
fig.append_trace(tr16, 2, 1)

fig.append_trace(tr13, 2, 2) # Palabras
fig.append_trace(tr14, 2, 2)

fig.append_trace(tr9, 3, 1)  # Mes
fig.append_trace(tr10, 3, 1)

fig.append_trace(tr7, 3, 2)  # Recursos
fig.append_trace(tr8, 3, 2)

# Actualizar el diseño general
fig.update_layout(
    barmode='group',
    title={
        'text': 'Comparación de Características Clave: <b>Proyectos con Financiación Rápida vs Lenta</b>',
        'font': {'size': 22, 'color': '#000'},
        'y': 0.98,
        'x': 0.5,
        'xanchor': 'center'
    },
    legend=dict(x=0.25, y=1.06, orientation="h"),
    margin=dict(t=100, b=100),
    paper_bgcolor='rgb(254, 247, 234)',
    plot_bgcolor='rgb(254, 247, 234)',
    height=1000
)

# Mostrar la figura
fig.show()
```
