```python
def bar_hor(df, col, title, color, w=None, h=None, lm=0, limit=100, return_trace=False, rev=False):
    cnt_srs = df[col].value_counts()
    yy = cnt_srs.head(limit).index[::-1]
    xx = cnt_srs.head(limit).values[::-1]

    if rev:
        yy = cnt_srs.tail(limit).index[::-1]
        xx = cnt_srs.tail(limit).values[::-1]

    trace = go.Bar(y=yy, x=xx, orientation='h',
                   marker=dict(color=color))

    if return_trace:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_hor_noagg(x, y, title, color, w=None, h=None, lm=0, limit=100, rt=False):
    trace = go.Bar(y=x, x=y, orientation='h',
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=0, rt=False):
    trace = go.Bar(y=y, x=x,
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


trace1 = bar_hor(
    donors_df,
    'Donor City',
    "Ciudades con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace2 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace4 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el menor número de donantes",
    '#c2d2ed',
    600, 400, 200,
    limit=10,
    return_trace=True,
    rev=True
)

fig = tools.make_subplots(
    rows=1,
    cols=2,
    print_grid=False,
    subplot_titles=[
        'Estados con el mayor número de donantes',
        'Estados con el menor número de donantes'
    ]
)

fig.append_trace(trace2, 1, 1)
fig.append_trace(trace4, 1, 2)

fig['layout'].update(height=400, showlegend=False)
iplot(fig)
```

## <a id="5.4"> 5.4 ¿Qué Tipos de Escuelas Están Registrados en Mayoría? </a>

```python
words = " ".join(schools_df['School Name']).split()
from collections import Counter

school_types = Counter(words).most_common(30)[1:9]
school_types = [list(x) for x in school_types]

# Diccionario de traducción de tipos de escuela
traduccion = {
    'Elementary': 'Escuela Primaria',
    'High': 'Escuela Preparatoria',
    'Middle': 'Escuela Secundaria',
    'Academy': 'Academia',
    'Charter': 'Escuela Charter',
    'Center': 'Centro Escolar',
    'Junior': 'Escuela Junior',
    'Senior': 'Escuela Senior'
}

# Usar traducción o mantener original si no está en el diccionario
labels = [traduccion.get(x[0], x[0]) for x in school_types]
values = [x[1] for x in school_types]

trace2 = go.Bar(
    x=labels,
    y=values,
    name='Cantidad de Escuelas',
    marker=dict(color='#ff00bb'),
    opacity=0.3
)

data = [trace2]
layout = go.Layout(
    barmode='group',
    legend=dict(dict(x=-.1, y=1.2)),
    title='Distribución de Tipos de Escuelas',
)

fig = go.Figure(data=data, layout=layout)
iplot(fig, filename='grouped-bar')
```

> Una gran mayoría de las escuelas son **Escuelas Primarias** (con aproximadamente **35K** escuelas), lo cual es mucho mayor que las **Escuelas Preparatorias** (aproximadamente **12K** escuelas), **Escuelas Secundarias** (aproximadamente **9K** escuelas) o **Academias Escolares** (aproximadamente **4K** escuelas).
